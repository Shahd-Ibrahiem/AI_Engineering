# **Day 1: LLM Engineering Foundations**

Objective: Setting up an OpenAI-Compatible environment using Groq

ℹ️ This notebook demonstrates the core setup for the AI Engineer track, focusing on:

Environment variable management with .env files.
Initializing an OpenAI-compatible client using Groq.
Creating a basic summarization pipeline for web content.

```python
import os
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display
from scraper import fetch_website_contents

# Initialize the environment
load_dotenv(override=True)

# Configuration for Groq (Our Free Alternative)
api_key = os.getenv('GROQ_API_KEY')
base_url = "https://api.groq.com/openai/v1"

# Client Setup
client = OpenAI(api_key=api_key, base_url=base_url)

# Validation
if not api_key:
    print("❌ Error: GROQ_API_KEY not found in .env file.")
else:
    print(f"✅ Environment Ready. Key starting with: {api_key[:8]}...")
```

### 1. Testing the Connection
We start by verifying that our client can communicate with the Groq servers. We will use llama-3.1-8b-instant for its high speed and efficiency.

```python
messages = [
    {"role": "system", "content": "You are a professional AI Engineer assistant."},
    {"role": "user", "content": "Confirm you are operational and briefly explain why Groq is fast."}
]

response = client.chat.completions.create(
    model="llama-3.1-8b-instant", 
    messages=messages
)

display(Markdown(f"**AI Response:**\n\n {response.choices[0].message.content}"))
```

### 2. Web Scraping & Summarization

Next, we define a pipeline to fetch content from a URL and generate a concise summary. This mimics a real-world RAG (Retrieval-Augmented Generation) workflow.

Note: Ensure scraper.py is in your project directory for the next cell to function.

```python
def summarize(url):
    # 1. Fetch
    content = fetch_website_contents(url)
    
    # 2. Prepare Prompts
    system_prompt = "You are an expert at summarizing technical websites."
    user_prompt = f"Please provide a concise, bulleted summary of the following website content:\n\n{content}"
    
    # 3. Request Summary
    response = client.chat.completions.create(
        model="llama-3.1-8b-instant",
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt}
        ]
    )
    return response.choices[0].message.content

# Test the function
target_url = "https://github.com/Shahd-Ibrahiem"
summary_result = summarize(target_url)

display(Markdown(f"### Summary for: {target_url}"))
display(Markdown(summary_result))
```
